# Banking System using Object-Oriented Programming

A Python-based banking system that demonstrates core Object-Oriented Programming (OOP) concepts through the simulation of bank branches, customers, accounts, transactions, and banking operations.

## OOP Concepts Demonstrated
- Classes and Objects
- Inheritance
- Abstraction
- Method Overriding
- Object Relationships
- Exception Handling

## Main Components
- Head Office
- Branches
- Customers
- Current and Savings Accounts
- Deposits and Withdrawals
- Account-to-Account Transfers
- Transaction History
- Account Charges

```python
from abc import ABC, abstractmethod
```

```python
 class HeadOffice:
    def __init__(self, address, bankName):
        self.address = address
        self.bankName = bankName
        self.branches = []

    def add_branch(self, branch):
        self.branches.append(branch)

    def list_branches(self):
        return self.branches
```

```python
class Branch:
    def __init__(self, address, manager):
        self.address = address
        self.manager = manager
        self.accounts = []

    def open_account(self, account):
        if account.branch != self:
            raise Exception("Account must be associated with this branch.")
        self.accounts.append(account)

    def get_account(self, accountNumber):
        for account in self.accounts:
            if account.accountNumber == accountNumber:
                return account
        return None

    def list_accounts(self):
        return self.accounts
```

```python
class Customer:
    def __init__(self, address, name):
        self.address = address
        self.name = name
        self.accounts = []

    def add_account(self, account):
        self.accounts.append(account)

    def list_accounts(self):
        return self.accounts
```

```python
class Account(ABC):
    def __init__(self, accountNumber, balance, branch, customer):
        self.accountNumber = accountNumber
        self.balance = balance
        self.branch = branch
        self.customer = customer
        self.transactions = []

    def deposit(self, amount):
        if amount <= 0:
            raise Exception("Deposit amount must be greater than 0.")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise Exception("Withdrawal amount must be greater than 0.")

        if amount > self.balance:
            raise Exception("Insufficient balance.")

        self.balance -= amount

    def addTransaction(self, tx):
        self.transactions.append(tx)

    @abstractmethod
    def calcCharges(self):
        pass
```

```python
class CurrentAccount(Account):
    def calcCharges(self):
        return 10
```

```python
class SavingsAccount(Account):
    def calcCharges(self):
        if self.balance < 5000:
            return 2
        return 0
```

```python
class Transaction:
    def __init__(self, value, debit_from, credit_to):
        self.value = value
        self.debit_from = debit_from
        self.credit_to = credit_to
```

```python
class BankService:
    @staticmethod
    def transfer(debit_from, credit_to, amount):
        if amount <= 0:
            raise Exception("Transfer amount must be greater than 0.")

        if debit_from == credit_to:
            raise Exception("Cannot transfer to the same account.")

        debit_from.withdraw(amount)
        credit_to.deposit(amount)

        tx = Transaction(amount, debit_from, credit_to)

        debit_from.addTransaction(tx)
        credit_to.addTransaction(tx)

        return tx
```

```python
# Create head office
head_office = HeadOffice("Makkah, Saudi Arabia", "Future Bank")

# Create branches
branch1 = Branch("Makkah", "Branch Manager 1")
branch2 = Branch("Jeddah", "Branch Manager 2")

head_office.add_branch(branch1)
head_office.add_branch(branch2)

# Create customers
customer1 = Customer("Makkah", "Ahad")
customer2 = Customer("Jeddah", "Sara")

# Create Customer 1 accounts
c1_current = CurrentAccount("C1-CA-1", 1000, branch1, customer1)
c1_savings = SavingsAccount("C1-SA-1", 5000, branch1, customer1)

customer1.accounts = [c1_current, c1_savings]

branch1.open_account(c1_current)
branch1.open_account(c1_savings)

# Create Customer 2 accounts
c2_current = CurrentAccount("C2-CA-1", 2000, branch2, customer2)
c2_savings = SavingsAccount("C2-SA-1", 6000, branch2, customer2)

customer2.accounts = [c2_current, c2_savings]

branch2.open_account(c2_current)
branch2.open_account(c2_savings)
```

```python
# Perform transactions
tx1 = BankService.transfer(c1_current, c2_savings, 300)
tx2 = BankService.transfer(c2_current, c1_savings, 500)

print("Transfer 1 completed:", tx1.value)
print("Transfer 2 completed:", tx2.value)
```

## System Demonstration

The following section demonstrates the banking system by creating branches, customers, and accounts, then performing different banking operations.

```python
# Create the head office
head_office = HeadOffice("Makkah, Saudi Arabia", "Future Bank")

# Create branches
branch1 = Branch("Makkah", "Branch Manager 1")
branch2 = Branch("Jeddah", "Branch Manager 2")

# Add branches to the head office
head_office.add_branch(branch1)
head_office.add_branch(branch2)

print("Bank:", head_office.bankName)
print("Number of branches:", len(head_office.list_branches()))
```

```python
# Create customers
customer1 = Customer("Makkah", "Ahad")
customer2 = Customer("Jeddah", "Sara")


# Create accounts for Customer 1
c1_current = CurrentAccount("C1-CA-1", 1000, branch1, customer1)
c1_savings = SavingsAccount("C1-SA-1", 5000, branch1, customer1)

customer1.accounts = [c1_current, c1_savings]

branch1.open_account(c1_current)
branch1.open_account(c1_savings)


# Create accounts for Customer 2
c2_current = CurrentAccount("C2-CA-1", 2000, branch2, customer2)
c2_savings = SavingsAccount("C2-SA-1", 6000, branch2, customer2)

customer2.accounts = [c2_current, c2_savings]

branch2.open_account(c2_current)
branch2.open_account(c2_savings)


print("Customer 1:", customer1.name)
print("Customer 2:", customer2.name)
print("Ahad's accounts:", len(customer1.accounts))
print("Sara's accounts:", len(customer2.accounts))
```

```python
# Perform transactions
tx1 = BankService.transfer(c1_current, c2_savings, 300)
tx2 = BankService.transfer(c2_current, c1_savings, 500)

print("Transfer 1 completed:", tx1.value)
print("Transfer 2 completed:", tx2.value)
```

```python
# Display final account information
all_accounts = [c1_current, c1_savings, c2_current, c2_savings]

print("----- ALL ACCOUNTS -----")

for acc in all_accounts:
    print(f"Account Number: {acc.accountNumber}")
    print(f"Customer Name: {acc.customer.name}")
    print(f"Branch Address: {acc.branch.address}")
    print(f"Balance: {acc.balance}")
    print(f"Number of Transactions: {len(acc.transactions)}")
    print("------------------------")
```

```python
# Display calculated charges for each account
print("----- ACCOUNT CHARGES -----")

for acc in all_accounts:
    print(f"Account {acc.accountNumber} charges = {acc.calcCharges()}")
```

```python
# Display all branches and their accounts
print("----- BRANCHES AND ACCOUNTS -----")

for branch in head_office.list_branches():
    print(f"Branch Address: {branch.address}, Manager: {branch.manager}")

    for acc in branch.list_accounts():
        print(
            f"   Account Number: {acc.accountNumber}, "
            f"Customer: {acc.customer.name}, "
            f"Balance: {acc.balance}"
        )
```
